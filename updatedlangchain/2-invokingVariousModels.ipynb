```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
```

```python
from langchain.chat_models import init_chat_model
geminimodel = init_chat_model("google_genai:gemini-3.7-flash")
response = geminimodel.invoke("Hello, how are you?")
response
```

### Streaming long responses### (Used in chatbot mainly)


```python
for chunk in geminimodel.stream("Tell me a story about a brave knight and a dragon in 200 words"):
    print(chunk.text)
```

Batching multiple requests

```python
geminimodel.batch([
    "Write a poem about the ocean.",
    "Explain the theory of relativity in simple terms.",
    "Generate a list of 10 healthy breakfast ideas."],
    config={'concurrency': 2, 'max_output_tokens': 100}
)
```
