Structured Ouput Using Pydantic


```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
```

```python
from langchain.chat_models import init_chat_model
model = init_chat_model("google_genai:gemini-3.7-flash")
model
```

Pydantic provides field validation,descriptions and nested structures.

```python
from pydantic import BaseModel, Field
class Movie(BaseModel):
    title: str = Field(description="The title of the movie")
    director: str = Field(description="The director of the movie")
    year: int = Field( description="The release year of the movie")
    genre: str = Field(description="The genre of the movie")
    rating: float = Field(description="The rating of the movie on a scale from 0 to 10")
```

```python
model_with_structure = model.with_structured_output(Movie)
model_with_structure
```

```python
response = model_with_structure.invoke("Can you provide details about the movie Inception?")
response
```

Nested Structure

```python
class Actor(BaseModel):
    name: str = Field(description="The name of the actor")
    birth_year: int = Field(description="The birth year of the actor")
    nationality: str = Field(description="The nationality of the actor")

class Movie(BaseModel):
    title: str = Field(description="The title of the movie")
    director: str = Field(description="The director of the movie")
    year: int = Field(description="The release year of the movie")
    genre: str = Field(description="The genre of the movie")
    rating: float = Field(description="The rating of the movie on a scale from 0 to 10")
    actors: list[Actor] = Field(description="A list of actors in the movie")    
```

```python
model_with_nested_structure = model.with_structured_output(Movie)
response = model_with_nested_structure.invoke("Can you provide details about the movie Inception")
response
```
